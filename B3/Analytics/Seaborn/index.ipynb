```python
Grades ={
    "subject":["English","English","Maths","Maths","Science","Science"],
    "grade":[85,90,78,88,92,85],
    "gender":["Female","Male","Female","Male","Female","Male"]
}
```

```python
df = pd.DataFrame(Grades)
```

```python
df
```

```python
#stacked bar plot
sns.barplot(x="subject", y="grade", hue="gender", data=df)
```
