To know the total number of dimesion of numpy array

```python
ar1 = np.array([21,54,89,93,59,90])
# print(ar1[1:])
print(ar1.ndim)
```

2D Array

```python
a2 = np.array([[12,34,67,89],[91,11,85,98]])
print(a2[0][1])
print(a2[1][2])
print(a2.ndim)
```

3D array

```python
a3 = np.array([  [[11,90,33],[41,65,32]],     [[12,65,98],[13,2,6]]  ])
print(a3[0][1][1])
print(a3[1][0][2])
print(a3[1][1][1])
print(a3.ndim)
```

**Statistical operation**

```python
aq = np.array([21,3,4,11])
np.sum(aq)  #Total sum
np.mean(aq)   #Average value
np.min(aq)   #minimum value
np.max(aq)  # maximum value
```
