```python
a2 = np.array([[12,34,67,89],[91,11,85,98]])
print(a2[0][1])
print(a2[1][2])
print(a2.ndim)
```

3D array

**Statistical operation**

```python
aq = np.array([21,3,4,11])
np.sum(aq)  #Total sum
np.mean(aq)   #Average value
np.min(aq)   #minimum value
np.max(aq)  # maximum value
```

**Mathematical operation**

```python
a = np.array([21,11,4,7,2])
b = np.array([8,5,22,9,10])
```

```python
np.add(a,b)
```

```python
np.subtract(a,b)
```

```python
np.multiply(a,b)
```

```python
np.divide(a,b)
```

```python
np.power(a,2)  #when we want to find square
np.power(a,3)  #when we want tofind cube
```

```python
np.sqrt(a)
```
