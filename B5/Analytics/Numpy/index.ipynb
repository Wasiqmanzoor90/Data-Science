**Data Manipulation**

```python
arr = np.array([1,2,3,4,5])
a1 = np.append(arr,[6,7,8,9,10])
print(a1)
```

```python
arr1 = np.array([[10,20,30],[40,50,60]])
a2 = np.append(arr1,[[70,80,90]],axis=0)
print(a2)
```

```python
a2[1,1] = 55
a2[2,2] = 99
a2[0,0] = 100
a2[0,2] = 35
a2[1,0] = 44
print(a2)
```

```python
#delete
a3 = np.delete(a2,2,axis=1)
print(a3)
```
