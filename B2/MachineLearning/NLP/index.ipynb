```python

# Text Representation in NLP
# BOW, CountVectorizer, TF-IDF

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer


# Sample text data
sentences = [
    "I love Python",
    "I love machine learning",
    "Python is easy"
]

# 1. BOW / CountVectorizer
# Converts text into vectors based on word counts
cv = CountVectorizer()
bow = cv.fit_transform(sentences)

print("1. BOW / CountVectorizer:")
print("Words:", cv.get_feature_names_out())
print(bow.toarray())


# 2. TF-IDF
# Gives higher importance to informative words
tfidf = TfidfVectorizer()
tfidf_result = tfidf.fit_transform(sentences)

print("\n2. TF-IDF:")
print(tfidf_result.toarray().round(2))
```
