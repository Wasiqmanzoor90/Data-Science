```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv(r"C:\Users\Dell\Documents\Project\Dataset\companies.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
lb = LabelEncoder()
#Label encoder converts data into numerical value because ML Model understands numerical values
data['Total_reviews'] = lb.fit_transform(data['Total_reviews'])
data['Avg_salary'] = lb.fit_transform(data['Avg_salary'])
data['Interviews_taken'] = lb.fit_transform(data['Interviews_taken'])
data['Total_jobs_available'] = lb.fit_transform(data['Total_jobs_available'])
data['Total_benefits'] = lb.fit_transform(data['Total_benefits'])
```

```python
data.head()
```

```python
#X here mean input data
x = data.iloc[:,[5,6,7,8,9]]
#Y here means output data
y = data['Ratings']
```

```python
#train test split
#random state is used to ensure the data is shuffled consistency and spit eac time
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=0)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
data.head()
```

```python
rating = pd.DataFrame([[343,522,172,198,379]],columns=x_test.columns)
pred = model.predict(rating)
print(f"Rating:{pred[0]:.1f}")
```
