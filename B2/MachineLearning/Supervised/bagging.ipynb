```python
import pandas as pd
```

```python
data = pd.read_csv(r"C:\Users\Dell\Documents\Project\Dataset\titanic.csv")
```

```python
data['Sex'].isnull().sum()
```

```python
data['Cabin']=data['Sex'].fillna('Unkown')
```

```python
from sklearn.preprocessing import LabelEncoder
lb = labelEncoder = LabelEncoder()
data['Sex'] = lb.fit_transform(data['Sex'])
data['Embarked'] = lb.fit_transform(data['Embarked'])
data['Cabin']=lb.fit_transform(data['Cabin'])
```

```python
x = data.iloc[:,[2,4,5,6,7,9,10,11]]
y = data['Survived']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.svm import SVC
```

```python

from sklearn.ensemble import BaggingClassifier

bagging_model = BaggingClassifier(
    estimator = DecisionTreeClassifier(),
    n_estimators = 30,
    random_state = 42
)
```

```python
bagging_model.fit(x_train,y_train)
bag_pred = bagging_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print('Bagging Accuracy',accuracy_score(y_test,bag_pred))
```
