```python
from sklearn.cluster import KMeans
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv(r"C:\Users\Dell\Documents\Project\Dataset\Iris.csv")
```

```python
data.head()
```

```python
#sum od square error
sse=[]
k_range = range(1,11)
for k in k_range:
    km = KMeans(n_clusters=k)
    km.fit(data[['SepalLengthCm','SepalWidthCm']])
    sse.append(km.inertia_)
plt.plot(k_range,sse,marker='o')
```
