```python
import pandas as pd
import seaborn as sns
```

```python
data = {
    "subject":['Math','Math','Science','Science','English','English'],
    "gender":['Male','Female','Male','Female','Male','Female'],
    "marks":[70,83,63,33,61,79]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
#stacked bar 
sns.barplot(x='subject',y='marks',data=df,hue='gender')
```
