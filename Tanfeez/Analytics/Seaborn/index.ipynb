```python
import seaborn as sns
import pandas as pd
```

```python
grades={
    "subject":['English','English','Math','Math','Science','Science'],
    "gender":['Male','Female','Male','Female','Male','Female',],
    "marks":[78,92,82,59,65,91]
    
}

data = pd.DataFrame(grades)
```

```python
data
```

```python
sns.barplot(x = 'subject',y='marks',hue='gender',data=data)
```
