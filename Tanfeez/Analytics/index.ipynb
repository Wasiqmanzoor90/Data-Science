```python
ar2 = np.array([21,45,98,11,10])
```

```python
np.max(ar2)
```

**Mathamatical operation**

```python
ar3 = np.array([2,4,6,8])
ar4 = np.array([10,12,14,16])
```

```python
np.add(ar3,ar4)
```

```python
np.subtract(ar3,ar4)
```

```python
np.multiply(ar3,ar4)
```

```python
np.divide(ar3,ar4)
```
