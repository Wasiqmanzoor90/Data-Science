```python
A = np.array([
    [1,2],
    [3,4]])
B = np.array([
    [5,6],
    [7,8]
    ])
```

```python
print("Matrix Addition\n", A@B)
#first row first colum
#1x5+2x7 = 5+14 = 19

#first second colum
#1x6+2x8 = 6 +16 = 22

#second row first colum
#3x5+4x7 = 15+28 = 43

#second row second colum
#3x6+4x8 = 18+32 = 50

```

```python
print("Matrix Addition\n", A+B)
```

```python
print("Matrix Element-wise Multiplication\n", A*B)
```
