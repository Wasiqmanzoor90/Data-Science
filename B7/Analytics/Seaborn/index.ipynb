```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
d = sns.load_dataset('titanic')
```

```python
d.head()
```

```python
d.shape
```

```python
grades={
    "Subject":["English","English",'Math','Math','SST','SST'],
    "Gender":['Male','Female','Male','Female','Male','Female'],
    'Marks':[78,92,83,65,59,61]
}
```

```python
d1 = pd.DataFrame(grades)
d1
```

```python
sns.barplot(x = 'Subject',y='Marks',hue='Gender',data=d1)
plt.xlabel("Subjects")
```
