**Array Manipulation**

```python
a1 = np.array([1,2,3,4])
arr = np.append(a1,[5,6,7,8])
print(arr)
```

```python
a2 = np.array([[10,20,30],[40,50,60]])
arr2 = np.append(a2,[[70,80,90]],axis=0)  #here 0 means row and 1 means column
print(arr2)
```

```python
arr2[1,0]=38
arr2[1,2] = 62
arr2[2,1]=87
arr2[0,2]=56
arr2[0,1]=21
print(arr2)
```

```python
a3 = np.delete(arr2,1,axis=1)
print(a3)
```
