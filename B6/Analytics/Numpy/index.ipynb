**Stastical operation**

```python
ar3 = np.array([1, 2, 3,21,34,45,11,76,10,11,76,10])
np.max(ar3)  #maximum value
np.mean(ar3)  #average value
np.min(ar3)  #minimum value
np.sum(ar3)  #sum of all item
np.median(ar3)  #middle value
```

**Mathamaitical operation**

```python
ar1 = np.array([2,4,6])
ar2 = np.array([8,9,10])
np.add(ar1,ar2)
```

```python
np.subtract(ar1,ar2)
```

```python
np.multiply(ar1,ar2)
```

```python
np.divide(ar1,ar2)
```
