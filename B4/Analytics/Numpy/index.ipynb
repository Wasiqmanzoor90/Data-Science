**Numpy array stores multiple items in a single variable. It's Homegenous**

```python
arr = np.array([2,56,89,11])
print(arr[1])
print(arr[-1])
print(arr[1:3])
print(arr[1:])
print(arr[:3])
print(arr[: :-1])
```

```python
ar1 = np.array([22,78,90,61,23])
print(ar1.ndim)
```

```python
#2D array
ar2 = np.array([[33,11,89,51,'wasiq'],[39,78,10,9,'umer']])
print(ar2)
print(ar2[0][2])
print(ar2[1][2])
```

```python
#3D array
ar3 = np.array([  [[21,45,89,22],[101,49,19,72]] ,       [[90,99,54,21],[1,5,69,32]]   ])
print(ar3[0][1][3])
print(ar3[0][0][2])
print(ar3[1][0][3])
```

**Mathematical Operation**

```python
a3 = np.array([21,34,79,10])
a4 = np.array([32,6,1,12])
```

```python
np.multiply(a3,a4)
```
