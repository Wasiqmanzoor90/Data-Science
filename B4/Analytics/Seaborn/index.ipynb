```python
import seaborn as sns
import pandas as pd
```

```python
grades = {
    "Subject":['Math','Math','Science','Science','History','History'],
    "Grade":[85, 90, 78, 82, 88, 91],
    "Gender":['Male','Female','Male','Female','Male','Female']
}
```

```python
data = pd.DataFrame(grades)
```

```python
sns.barplot(x='Subject', y='Grade',hue='Gender',data=data)
```
