**Numerical Python also know as numpy was created by scientist namely travis oliphan in 2005 his main motive was to reduce the time complexity so thats why he create a numpy array which basically works like normal python list but it's 50x faster than normal python list**

```python
#Here we inmport numpy
import numpy as np
```

**Numpy array stores multiple items in a single variable. It's Homegenous**

```python
arr = np.array([2,56,89,11])
print(arr[1])
print(arr[-1])
print(arr[1:3])
print(arr[1:])
print(arr[:3])
print(arr[: :-1])
```

```python
ar1 = np.array([22,78,90,61,23])
print(ar1.ndim)
```

```python
#2D array
ar2 = np.array([[33,11,89,51,'wasiq'],[39,78,10,9,'umer']])
print(ar2)
print(ar2[0][2])
print(ar2[1][2])
```

```python
#3D array
ar3 = np.array([  [[21,45,89,22],[101,49,19,72]] ,       [[90,99,54,21],[1,5,69,32]]   ])
print(ar3[0][1][3])
print(ar3[0][0][2])
print(ar3[1][0][3])
```
